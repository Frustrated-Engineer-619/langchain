#### Embedding Techniques

Convert Text into vectors

```python
import os
from dotenv import load_dotenv
load_dotenv() #load all the environment variables from .env file
```

```python
os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
```

```python
from langchain_openai import OpenAIEmbeddings
embedding = OpenAIEmbeddings(model="text-embedding-3-large")
```

```python
text="This is a tutorial on OPENAI embeddings."
query_result = embedding.embed_query(text)
query_result
```

```python
len(query_result)
```

```python
embedding_1024 = OpenAIEmbeddings(model="text-embedding-3-large",dimensions=1024)
```

```python
text="This is a tutorial on OPENAI embeddings."
query_result = embedding_1024.embed_query(text)
len(query_result)
```

```python
from langchain_community.document_loaders import TextLoader
loader = TextLoader('speech.txt')
docs = loader.load()
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
text_splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=50)
final_docs = text_splitter.split_documents(docs)
final_docs
```

```python
## Vector Enbedding and Storing in Vector Database
from langchain_community.vectorstores import Chroma
db = Chroma.from_documents(final_docs, embedding_1024)
db
```

```python
query = "It will be all the easier for us to conduct ourselves "
retrieved_results = db.similarity_search(query)

print(retrieved_results)
```
