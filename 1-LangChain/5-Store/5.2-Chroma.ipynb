## Chroma

Chroma is a AI-natice open-source vector database focused on developer productivity and happiness. Chroma is licensed under Apache 2.0

```python
from langchain_chroma import Chroma
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.embeddings import OllamaEmbeddings
```

```python
## Data Ingestion
loader = TextLoader("speech.txt")
data = loader.load()
data
```

```python
## Text Splitting : Splitting the documents to chunks of 500 characters with an overlap of 30 characters
text_splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=30)
docs = text_splitter.split_documents(data)
```

```python
### Embeddings : Creating embeddings for the documents using OllamaEmbeddings
embeddings = OllamaEmbeddings(model="nomic-embed-text")


## takes your documents, converts them into vectors using the embedding 
# model, and stores those vectors in a Chroma vector database.
vectordb = Chroma.from_documents(docs, embeddings)
```

```python
##Querying the database for similar documents based on a query
query = "What is the main topic of the speech?"
docs = vectordb.similarity_search(query)
docs
```

```python
##Saving to the disk
vectordb = Chroma.from_documents(docs, embeddings,persist_directory="./chroma_db")
```

```python
# Load from Disk

db2 = Chroma(persist_directory="./chroma_db", embedding_function=embeddings)
docs = db2.similarity_search(query)
print(docs[0].page_content)
```

```python
#Similarity Search with Score
docs = db2.similarity_search_with_score(query)
print(docs)
for doc, score in docs:
    print(f"Score: {score}, Document: {doc.page_content}")
```

```python
##Retriever

retriever = db2.as_retriever()
retriever.invoke(query)[0]
```
