```python
# ChatpromptTemplate
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages(
    [
        ("system", "You are an expert AI engineer.Provider me answers based on the questions"),
        ("user","{input}")
    ]
)

prompt
```

```python
# Its basically saying that we need to combine this prompt along  with my llm and that actually becomes a chain.
# Now whenever I give any input, first of all it will go through the prompt and then it will go to the llm and get aresponse and then it will give me the final output.
chain = prompt | llm

response = chain.invoke({"input":"can you tell me about LangSmith?"})
print(response)
```

```python
## stroutput parser

from langchain_core.output_parsers import StrOutputParser
output_parser = StrOutputParser()
chain = prompt | llm | output_parser

response = chain.invoke({"input":"can you tell me about hallucination?"})
print(response)
```
